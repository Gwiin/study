```python
def train_test_split(a):
    divisor = int(len(a)*0.8)
    train_data = a[:divisor]
    test_data = a[divisor:]
    return (train_data, test_data)
a = np.arange(1,51)
np.random.shuffle(a)
a = np.reshape(a,(5,-1))
(train_data, test_data) = train_test_split(a)
print(f'Train : {train_data}, \n Test : {test_data}')
```

```python
arr_a = np.random.rand(1000)
arr_b = np.random.rand(1000)
result = np.zeros(1000)
```

```python
start = time.time()
for i in range(len(arr_a)):
    result[i] = arr_a[i] * arr_b[i]
end = time.time()
print('소요시간: ', end - start)
```
