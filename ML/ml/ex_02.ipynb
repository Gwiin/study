```python
from sklearn import linear_model
```

```python
reg = linear_model.LinearRegression()
```

```python
X = [[0],[1],[2]]
y = [[3],[4],[5]]
```

```python
reg.fit(X,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
reg.predict([[-4]])
```

```python
reg.predict([[5]])
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X,y, color='green')
plt.plot(X, reg.predict(X), c = 'purple', linewidth=2)
plt.show()
```
