```python
import numpy as np
```

```python
x = np.arange(-10,11)
y = np.square(x)
# y =  x ** 2
plt.plot(x,y)
plt.axis([-100,100,0,100])
# plt.stem(x,y)
plt.show()
```

```python
x = np.arange(-20,21)
y1 = x * 2
y2 = (1/3) * np.square(x) + 5
y3 = -(x ** 2) - 5
plt.plot(x, y1, 'g--')
plt.plot(x, y2, 'b-*')
plt.plot(x, y3, 'r.:')
plt.axis([-30,30,-30,30])
plt.show()
```

```python
x = np.arange(50)
y = np.random.random(50)
plt.plot(x,y,'b:o')
plt.show()
```

```python
x = np.linspace(0,2 * np.pi, 1000)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x,y1,'b--', label = "Sine wave")
plt.plot(x,y2,'r--', label = "Cosine wave")
plt.legend()
plt.xlabel("time")
plt.ylabel("magnitude")
plt.savefig("Sine-Cosine.png")
plt.show()
```

```python
fig, ax = plt.subplots(2,2)

X = np.random.randn(100)
Y = np.random.randn(100)
ax[0,0].scatter(X, Y)
X = np.arange(10)
Y = np.random.uniform(1, 10, 10)
ax[0,1].bar(X, Y)
X = np.linspace(0, 10, 100)
Y = np.cos(X)
ax[1,0].plot(X,Y)
Z = np.random.uniform(0, 1, (5,5))
ax[1,1].imshow(Z)
```

```python
fig = plt.figure(figsize=(7, 5), constrained_layout=True)
grid = plt.GridSpec(3, 3, figure=fig)

X = np.random.randn(100)
Y = np.random.randn(100)
ax1 = fig.add_subplot(grid[0, :])
ax1.scatter(X, Y)

X = np.linspace(0, 10, 100)
Y = np.linspace(0, 10, 100)
ax2 = fig.add_subplot(grid[1, 0:2])
ax2.plot(X, Y)

X = np.linspace(0, 10, 100)
Y = np.cos(X)
ax3 = fig.add_subplot(grid[2, 0])
ax3.plot(X, Y)

X = np.arange(10)
Y = np.random.uniform(1, 10, 10)
ax4 = fig.add_subplot(grid[2, 1])
ax4.bar(X, Y)

Z = np.random.uniform(0, 1, (10, 5))
ax5 = fig.add_subplot(grid[1:, 2])
ax5.imshow(Z, aspect="auto")

plt.show()
```

```python
heights = np.array(np.random.randint(150,180,30))
plt.hist(heights, bins = 6)
plt.xlabel("height")
plt.ylabel("frequency")
```

```python
plt.hist(heights, bins=6, label='cumulative=True', cumulative=True)
plt.hist(heights, bins=6, label='cumulative=False', cumulative=False)
plt.legend(loc='upper left')
```

```python
f1 = np.random.randn(100_000)
plt.hist(f1, bins=200, color = "red", alpha = .7, label = 'avg = 0, std = 1')
plt.axis([-8,8,-2,2500])
plt.legend()
```

```python
f1 = np.random.normal(loc=0, scale=1, size=100000)
f2 = np.random.normal(loc=3, scale=1, size=100000)

plt.hist(f1, bins=200, color = "red", alpha = .4, label = 'avg = 0, std = 1')
plt.hist(f2, bins=200, color = "green", alpha = .4, label = 'avg = 3, std = 1')
plt.axis([-8,8,-2,2500])
plt.legend()
```
